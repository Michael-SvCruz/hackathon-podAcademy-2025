### Silver Bureau

```python
path_silver = "/Volumes/hackathon_2025/default/silver/bureau_full_silver_delta/"
```

```python
df_silver = spark.read.format("delta").load(path_silver)

display(df_silver)
```

```python
%sql
SELECT *
FROM hackathon_2025.default.silver_bureau
LIMIT 5
```

```python
%sql
SELECT
  COUNT(*) AS total,
  COUNT(DISTINCT CONCAT(NUM_CPF,'#',SAFRA)) AS distinct_key,
  COUNT(*) - COUNT(DISTINCT CONCAT(NUM_CPF,'#',SAFRA)) AS duplicadas
FROM hackathon_2025.default.silver_bureau;
```

```python
%sql
SELECT FLAG_INSTALACAO_INT
, COUNT(*) AS qtd
FROM hackathon_2025.default.silver_bureau
GROUP BY 1
ORDER BY qtd DESC;
```

```python
%sql
SELECT
  SUM(CASE WHEN FLAG_SCORE01_MISSING = 1 THEN 1 ELSE 0 END) AS score01_missing,
  COUNT(*) AS total
FROM hackathon_2025.default.silver_bureau;
```

```python
%sql
SELECT
  FLAG_INSTALACAO_INT AS flag_instalacao,
  FPD_INT AS fpd,
  COUNT(*) AS qtd
FROM hackathon_2025.default.silver_bureau
GROUP BY 1,2
ORDER BY 1,2;
```

```python
%sql
CREATE OR REPLACE TABLE hackathon_2025.default.gold_spine AS
  SELECT
    NUM_CPF,
    SAFRA,
    DT_SAFRA,
    FLAG_INSTALACAO_INT,
    FPD_INT,
    SCORE_01_ADJ,
    FLAG_SCORE01_MISSING
  FROM hackathon_2025.default.silver_bureau
```

```python
%sql
SELECT
  NUM_CPF,
  SAFRA,
  DT_SAFRA,
  FLAG_INSTALACAO_INT,
  FPD_INT,
  SCORE_01_ADJ,
  FLAG_SCORE01_MISSING
FROM hackathon_2025.default.silver_bureau;
```

```python
%sql
SELECT * FROM hackathon_2025.default.gold_spine
```
