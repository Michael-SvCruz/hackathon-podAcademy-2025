#Checks

```python
%sql
SELECT Safra,  
      COUNT(*)
        
FROM telco
GROUP BY 1
ORDER BY 1
```

```python
%sql
SELECT 
      FPD,  
      COUNT(*)
        
FROM telco
GROUP BY ALL
ORDER BY ALL
```

```python
%sql
SELECT Safra,
      FPD,  
      COUNT(*)
        
FROM telco
GROUP BY ALL
ORDER BY ALL
```

```python
%sql
SELECT Safra,
      ROUND(SUM(FPD)/COUNT(*),4) AS pct_FPD,  
      COUNT(*)
        
FROM telco
WHERE FPD IS NOT NULL
GROUP BY ALL
ORDER BY ALL
```

```python
%sql
SELECT * FROM telco
WHERE FPD IS NULL
```

```python
%sql
SELECT * FROM telco
WHERE FPD IS NULL AND flag_mig2 IS NOT NULL
```

```python
%sql
SELECT * FROM telco
WHERE FPD IS NULL AND FLAG_INSTALACAO != 0
```

```python
%sql
SELECT * FROM telco
WHERE FPD IS NOT NULL AND FLAG_INSTALACAO != 1
```

```python
%sql
SELECT Safra,  
      COUNT(*)
        
FROM telco
WHERE FPD IS NOT NULL AND flag_mig2 = "PRE"
GROUP BY 1
ORDER BY 1
```

```python
%sql
SELECT Safra,
      flag_mig2,
      ROUND(SUM(FPD)/COUNT(*),4) AS pct_target,  
      COUNT(*)
        
FROM telco
WHERE FPD IS NOT NULL
GROUP BY ALL
ORDER BY ALL
```
